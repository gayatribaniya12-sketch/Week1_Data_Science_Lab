# Real Business Problem

## Online Store Customer Demand

Problem:
An online store does not know how many products customers will buy.

Data Science Solution:
The company can use previous sales data to predict future customer demand.

Benefits:
- Reduce product waste
- Avoid products being out of stock
- Improve inventory management
- Increase customer satisfaction

## Raw Data and Business Questions

### Raw Data

An online store might collect:
- Product name and category
- Number of products sold
- Sales date and time
- Product price
- Customer orders
- Customer location
- Stock/inventory level
- Discounts and promotions
- Customer ratings and reviews

### Business Questions

This data can answer:
1. Which products are selling the most?
2. Which products are selling the least?
3. How many products might customers buy next month?
4. When is customer demand highest?
5. Which products need more inventory?
6. Do discounts increase sales?
